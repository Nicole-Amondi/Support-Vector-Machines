```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Movie_classification.csv', header=0)
df.head()
```

```python
df.info()
```

Missing Value Imputation

```python
df['Time_taken'].mean()
```

```python
df['Time_taken'] = df['Time_taken'].fillna(df['Time_taken'].mean())
df.info()
```

Dummy Variable Creation

```python
df.head()
```

```python
df = pd.get_dummies(df,columns = ['3D_available', 'Genre'], drop_first=True)
df[['3D_available_YES', 'Genre_Comedy', 'Genre_Drama', 'Genre_Thriller']] = df[['3D_available_YES', 'Genre_Comedy', 'Genre_Drama', 'Genre_Thriller']].astype(int)
```

```python
df.head()
```

X-y Split

```python
X = df.loc[:, df.columns != 'Start_Tech_Oscar']
type(X)
```

```python
X.head()
```

```python
y = df['Start_Tech_Oscar']
type(y)
```

```python
y.head()
```

Test-Train Split

```python
!pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.2, random_state=0)
```

```python
X_train.shape
```

```python
X_test.shape
```

Standardizing Data

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler().fit(X_train)
```

```python
X_train_std = sc.transform(X_train)
X_test_std = sc.transform(X_test)
```

```python
X_test_std
```

SVM Classification Model

Training Classification Tree

```python
from sklearn import svm
```

```python
clf_svm_1 = svm.SVC(kernel='linear', C=0.01)
clf_svm_1.fit(X_train_std, y_train)
```

Predict Values Using Trained Model

```python
y_train_pred = clf_svm_1.predict(X_train_std)
y_test_pred = clf_svm_1.predict(X_test_std)
```

```python
y_test_pred
```

Model Performance

```python
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
confusion_matrix(y_test, y_test_pred)
```

```python
accuracy_score(y_test, y_test_pred)
```
